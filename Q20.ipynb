```python
import os

print("Searching for PlantDisease folder...\n")

locations = [
    os.path.expanduser("~/Desktop"),
    os.path.expanduser("~/Downloads"),
    os.path.expanduser("~/Documents")
]

found = False

for location in locations:
    if os.path.exists(location):
        for root, dirs, files in os.walk(location):
            if "PlantDisease" in dirs:
                path = os.path.join(root, "PlantDisease")
                print("✅ Dataset found here:")
                print(path)
                found = True
                break

    if found:
        break

if not found:
    print("❌ PlantDisease folder was not found.")
    print("\nPlease check where you extracted/downloaded the dataset.")
```
