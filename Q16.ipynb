```python
# Rule-Based Expert System for Student Performance Evaluation

print("STUDENT PERFORMANCE EVALUATION SYSTEM")

# Taking input from the user
attendance = float(input("Enter attendance percentage: "))
marks = float(input("Enter marks percentage: "))
assignment = float(input("Enter assignment performance percentage: "))
internal = float(input("Enter internal assessment percentage: "))

# IF-THEN rules
if attendance >= 90 and marks >= 90 and assignment >= 90 and internal >= 90:
    result = "Excellent"

elif attendance >= 75 and marks >= 75 and assignment >= 75 and internal >= 75:
    result = "Good"

elif attendance >= 60 and marks >= 60 and assignment >= 60 and internal >= 60:
    result = "Average"

else:
    result = "Poor"

# Display the result
print("Student Performance:", result)
```
