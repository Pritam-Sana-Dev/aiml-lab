```python
from collections import deque
import networkx as nx
import matplotlib.pyplot as plt

# Define the graph
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['G'],
    'F': [],
    'G': []
}

# DFS Function
def dfs(graph, start):
    visited = set()
    traversal = []

    def dfs_recursive(node):
        visited.add(node)
        traversal.append(node)

        for neighbor in graph[node]:
            if neighbor not in visited:
                dfs_recursive(neighbor)

    dfs_recursive(start)
    return traversal

# Run DFS
result = dfs(graph, 'A')

print("DFS Traversal:")
print("->".join(result))

# Create directed graph
G = nx.DiGraph()

# Add edges from the dictionary
for node in graph:
    for neighbor in graph[node]:
        G.add_edge(node, neighbor)

# Define positions for better tree layout
pos = {
    'A': (0, 3),
    'B': (-1, 2),
    'C': (1, 2),
    'D': (-1.5, 1),
    'E': (-0.5, 1),
    'F': (1, 1),
    'G': (-0.5, 0)
}

# Draw the graph
plt.figure(figsize=(8,6))

nx.draw(
    G, pos,
    with_labels=True,
    node_color='lightgreen',
    node_size=2500,
    font_size=14,
    font_weight='bold',
    edge_color='black',
    arrows=True,
    width=2
)

plt.title("DFS Graphical Representation", fontsize=16)
plt.axis('off')
plt.show()
```
