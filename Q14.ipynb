```python
# Minimax Algorithm for Two-Player Game

def minimax(tree, maximizing):
    # If it is a leaf node
    if isinstance(tree, int):
        return tree

    scores = []

    # Recursively evaluate all child nodes
    for child in tree:
        scores.append(minimax(child, not maximizing))

    # MAX player chooses maximum score
    if maximizing:
        return max(scores)

    # MIN player chooses minimum score
    else:
        return min(scores)


# Game tree using given leaf-node values
tree = [
    [[3, 5], [2, 9]],
    [[12, 5], [23, 23]]
]

# MAX player starts
optimal_score = minimax(tree, True)

print("Leaf-node values:", 3, 5, 2, 9, 12, 5, 23, 23)
print("Optimal score for MAX player:", optimal_score)
```
