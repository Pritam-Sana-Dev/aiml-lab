```python
# Alpha-Beta Pruning for Two-Player Game

def alpha_beta(depth, index, maximizing, values, alpha, beta, max_depth):

    # Leaf node
    if depth == max_depth:
        return values[index]

    if maximizing:
        best = -float("inf")

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                index * 2 + i,
                False,
                values,
                alpha,
                beta,
                max_depth
            )

            best = max(best, value)
            alpha = max(alpha, best)

            # Beta cutoff
            if beta <= alpha:
                print("Pruned branch at index:", index * 2 + i)
                break

        return best

    else:
        best = float("inf")

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                index * 2 + i,
                True,
                values,
                alpha,
                beta,
                max_depth
            )

            best = min(best, value)
            beta = min(beta, best)

            # Alpha cutoff
            if beta <= alpha:
                print("Pruned branch at index:", index * 2 + i)
                break

        return best


# Given leaf-node values
values = [3, 5, 2, 9, 12, 5, 23, 23]

# Tree depth = 3
max_depth = 3

# Start Alpha-Beta Pruning
score = alpha_beta(
    0,
    0,
    True,
    values,
    -float("inf"),
    float("inf"),
    max_depth
)

print("Leaf-node values:", values)
print("Optimal score for MAX:", score)
```
